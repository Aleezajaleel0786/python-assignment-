```
# 1. Create a list and add up to 5 different numbers
grades = [85, 90, 78, 88, 95]  # Example list of grades

# 2. Add grade 95 to the list
grades.append(95)

# 3. Calculate the average grade
average_grade = sum(grades) / len(grades)

# Print the average grade using an f-string
print(f"Average Grade: {average_grade:.2f}")

# 4. Find the highest and lowest grade
highest = max(grades)
lowest = min(grades)

# Print highest and lowest grade using f-strings
print(f"Highest Grade: {highest}")
print(f"Lowest Grade: {lowest}")
```
